#### CHECKING MISSING VALUES

```python
missing_summary = pd.DataFrame({
    "missing_count": df.isnull().sum(),
    "missing_percentage": (df.isnull().mean() * 100).round(2),
    "non_missing_count": df.notnull().sum()
})

missing_summary = missing_summary.sort_values(
    by="missing_percentage",
    ascending=False
)

display(missing_summary)
```

### Test date conversion

```python
# Date parsing validation

date_columns = [
    "created_date",
    "closed_date",
    "due_date",
    "resolution_action_updated_date"
]

for col in date_columns:
    parsed = pd.to_datetime(df[col], errors="coerce")

    invalid_dates = df[col].notna() & parsed.isna()

    print(f"\n{col}")
    print("-" * 40)
    print(f"Original non-null: {df[col].notna().sum()}")
    print(f"Successfully parsed: {parsed.notna().sum()}")
    print(f"Invalid dates: {invalid_dates.sum()}")

    if invalid_dates.any():
        print("Examples:", df.loc[invalid_dates, col].head().tolist())
```

### Temporal consistency checks


```python
date_columns = [
    "created_date",
    "closed_date",
    "due_date",
    "resolution_action_updated_date"
]

# Convert into a separate temporary DataFrame
dates = df[date_columns].apply(
    pd.to_datetime,
    errors="coerce"
)

checks = {
    "closed_before_created": (
        dates["closed_date"] < dates["created_date"]
    ),
    "due_before_created": (
        dates["due_date"] < dates["created_date"]
    ),
    "resolution_update_before_created": (
        dates["resolution_action_updated_date"]
        < dates["created_date"]
    )
}

for name, condition in checks.items():
    print(f"{name}: {condition.sum()}")
```

```python
# Records closed before creation
closed_invalid = df.loc[
    dates["closed_date"] < dates["created_date"],
    ["unique_key", "created_date", "closed_date"]
].copy()

print("CLOSED BEFORE CREATED")
print("-" * 50)
display(closed_invalid)


# Records with resolution update before creation
update_invalid = df.loc[
    dates["resolution_action_updated_date"]
    < dates["created_date"],
    [
        "unique_key",
        "created_date",
        "resolution_action_updated_date",
        "status",
        "complaint_type"
    ]
].copy()

update_invalid["time_difference_hours"] = (
    dates.loc[update_invalid.index, "resolution_action_updated_date"]
    - dates.loc[update_invalid.index, "created_date"]
).dt.total_seconds() / 3600
```

```python
print("\nRESOLUTION UPDATE BEFORE CREATED")
print("-" * 50)
print(f"Total records: {len(update_invalid)}")

display(update_invalid.head(15))

print("\nTime difference summary (hours):")
display(update_invalid["time_difference_hours"].describe())
```
